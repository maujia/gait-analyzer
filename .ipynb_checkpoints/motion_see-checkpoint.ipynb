```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from scipy.signal import find_peaks, savgol_filter
```

```python
# --- load data from text file ---
def is_data_line(s):
    return not s.startswith("RAW")
t_ms, back_deg, thigh_deg, rel_deg = np.loadtxt(
    (line for line in open("data2.txt") if is_data_line(line)),
    unpack=True, dtype=float
)

# Convert time to seconds
t_sec = t_ms / 1000.0

# Calculate ROM metrics
max_flexion = np.max(rel_deg)
max_extension = np.min(rel_deg)
total_rom = max_flexion - max_extension
mean_angle = np.mean(rel_deg)
total_time = len(rel_deg)
time_flexed = np.sum(rel_deg > 0) / total_time * 100
time_extended = np.sum(rel_deg < 0) / total_time * 100

# Create single plot
fig, ax1 = plt.subplots(figsize=(14, 7))

# Plot line
ax1.plot(t_sec, rel_deg, 'b-', linewidth=1.5, label='Hip Flexion')

# Highlight flexion regions
ax1.fill_between(t_sec, 0, rel_deg, where=(rel_deg>0), color='lightgreen', alpha=0.3, label='Flexion')
ax1.fill_between(t_sec, 0, rel_deg, where=(rel_deg<0), color='lightcoral', alpha=0.3, label='Extension')

# Mark max/min points
max_idx = np.argmax(rel_deg)
min_idx = np.argmin(rel_deg)
ax1.plot(t_sec[max_idx], max_flexion, 'go', markersize=10, label=f'Max Flexion: {max_flexion:.1f}°')
ax1.plot(t_sec[min_idx], max_extension, 'ro', markersize=10, label=f'Max Extension: {max_extension:.1f}°')

# Reference lines
ax1.axhline(y=0, color='k', linestyle='--', alpha=0.5, linewidth=1)
ax1.axhline(y=mean_angle, color='purple', linestyle=':', alpha=0.7, linewidth=2, label=f'Mean: {mean_angle:.1f}°')

"""
# Add text box with ROM stats
textstr = f'ROM Analysis:\n' \
          f'Total ROM: {total_rom:.1f}°\n' \
          f'Time Flexed: {time_flexed:.1f}%\n' \
          f'Time Extended: {time_extended:.1f}%'
ax1.text(0.02, 0.98, textstr, transform=ax1.transAxes, fontsize=11,
         verticalalignment='top', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))
"""

ax1.set_xlabel('Time (seconds)', fontsize=12)
ax1.set_ylabel('Hip Flexion Angle (degrees)', fontsize=12)
ax1.set_title('Hip Flexion ROM Analysis', fontsize=14, fontweight='bold')
ax1.grid(True, alpha=0.3)
ax1.legend(loc='upper right')

plt.tight_layout()
plt.show()
```

```python
from scipy.signal import find_peaks

# Load data
def is_data_line(s):
    return not s.startswith("RAW") and not s.startswith("I (")

t_ms, back_deg, thigh_deg, rel_deg = np.loadtxt(
    (line for line in open("data.txt") if is_data_line(line)),
    unpack=True, dtype=float, usecols=(0, 1, 2, 3)
)
t_sec = t_ms / 1000.0

# Find flexion peaks and calculate metrics
peaks, _ = find_peaks(rel_deg, height=50, distance=100)
flexion_durations = []
time_to_peak = []

for peak_idx in peaks:
    start_idx = peak_idx
    for i in range(peak_idx - 1, -1, -1):
        if rel_deg[i] <= 0:
            start_idx = i
            break
    
    peak_angle = rel_deg[peak_idx]
    end_idx = peak_idx
    for i in range(peak_idx + 1, len(rel_deg)):
        if rel_deg[i] < peak_angle * 0.7:
            end_idx = i
            break
    
    flexion_durations.append(t_sec[end_idx] - t_sec[start_idx])
    time_to_peak.append(t_sec[peak_idx] - t_sec[start_idx])

# Plot
fig, ax1 = plt.subplots(figsize=(14, 7))
ax1.plot(t_sec, rel_deg, 'b-', linewidth=1.5, label='Hip Flexion')
ax1.fill_between(t_sec, 0, rel_deg, where=(rel_deg>0), color='lightgreen', alpha=0.3, label='Flexion')
ax1.axhline(y=0, color='k', linestyle='--', alpha=0.5, linewidth=1)

# Stats box
textstr = f'Flexion Events: {len(peaks)}\n' \
          f'Avg Duration (lift+hold): {np.mean(flexion_durations):.2f}s\n' \
          f'Avg Time to Peak: {np.mean(time_to_peak):.2f}s'
ax1.text(0.02, 0.98, textstr, transform=ax1.transAxes, fontsize=11,
         verticalalignment='top', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8))

ax1.set_xlabel('Time (seconds)', fontsize=12)
ax1.set_ylabel('Hip Flexion Angle (degrees)', fontsize=12)
ax1.set_title('Hip Flexion Analysis', fontsize=14, fontweight='bold')
ax1.grid(True, alpha=0.3)
ax1.legend(loc='upper right')
plt.tight_layout()
plt.show()
```

```python
# Load data
def is_data_line(s):
    return (not s.startswith("RAW")) and (not s.startswith("I (")) and (s.strip() != "")

t_ms, back_deg, thigh_deg, rel_deg = np.loadtxt(
    (line for line in open("data.txt") if is_data_line(line)),
    unpack=True, dtype=float, usecols=(0, 1, 2, 3)
)
t_sec = t_ms / 1000.0

# -----------------------------
# BEST FIX: unwrap + smooth
# -----------------------------
# Unwrap removes the +180 to -180 "flip" artifacts that create fake peaks.
rel_unwrapped = np.rad2deg(np.unwrap(np.deg2rad(rel_deg)))

# Light smoothing to reduce jitter that can split one rep into multiple peaks.
# Window length must be odd and should be tuned to your sample rate.
rel_smooth = savgol_filter(rel_unwrapped, window_length=51, polyorder=3)

# Peak detection (tune distance/prominence if needed)
peaks, props = find_peaks(
    rel_smooth,
    height=50,
    distance=300,
    prominence=20
)

# Compute metrics: duration (lift+hold) and time-to-peak
flexion_durations = []
time_to_peak = []

for peak_idx in peaks:
    # start = last time signal was at/below 0 before peak
    start_idx = peak_idx
    for i in range(peak_idx - 1, -1, -1):
        if rel_smooth[i] <= 0:
            start_idx = i
            break

    peak_angle = rel_smooth[peak_idx]

    # end = first time after peak when it falls below 70% of peak
    end_idx = peak_idx
    for i in range(peak_idx + 1, len(rel_smooth)):
        if rel_smooth[i] < peak_angle * 0.7:
            end_idx = i
            break

    flexion_durations.append(t_sec[end_idx] - t_sec[start_idx])
    time_to_peak.append(t_sec[peak_idx] - t_sec[start_idx])

# Plot
fig, ax1 = plt.subplots(figsize=(14, 7))
ax1.plot(t_sec, rel_smooth, 'b-', linewidth=1.5, label='Hip Flexion (unwrapped+smoothed)')
ax1.fill_between(t_sec, 0, rel_smooth, where=(rel_smooth > 0), color='lightgreen', alpha=0.3, label='Flexion')
ax1.axhline(y=0, color='k', linestyle='--', alpha=0.5, linewidth=1)

# Mark peaks
ax1.plot(t_sec[peaks], rel_smooth[peaks], 'ro', markersize=6, label='Detected Peaks')

# Stats box
avg_dur = np.mean(flexion_durations) if flexion_durations else float('nan')
avg_ttp = np.mean(time_to_peak) if time_to_peak else float('nan')

textstr = (
    f'Flexion Events: {len(peaks)}\n'
    f'Avg Duration (lift+hold): {avg_dur:.2f}s\n'
    f'Avg Time to Peak: {avg_ttp:.2f}s'
)
ax1.text(
    0.02, 0.98, textstr,
    transform=ax1.transAxes, fontsize=11,
    verticalalignment='top',
    bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.8)
)

ax1.set_xlabel('Time (seconds)', fontsize=12)
ax1.set_ylabel('Hip Flexion Angle (degrees)', fontsize=12)
ax1.set_title('Hip Flexion Analysis', fontsize=14, fontweight='bold')
ax1.grid(True, alpha=0.3)
ax1.legend(loc='upper right')
plt.tight_layout()
plt.show()
```
